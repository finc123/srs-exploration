# Roblox: Sensor Tower DAU vs our Consistent CCU, YTD 2026

**Question:** do Sensor Tower's daily active users (DAU) and our Consistent-panel CCU move together? Periods where one moves and the other doesn't could point to changes in bot activity, since bots add concurrent players without adding real app users. The shaded band marks the **17–19 Aug 2026 bot clampdown**.

**Sources**
- **DAU:** Sensor Tower in Snowflake, `SENSORTOWER.COMMON.CORE`: the worldwide (`WW`) row for the Roblox and Roblox VN apps, iOS + Android, daily from 2018. The `WW` row equals the sum of Sensor Tower's 80 tracked countries; the table also has one row per country, so only `WW` is used, to avoid double counting. **Mobile apps only:** no PC, Mac, console or browser.
- **CCU:** `SRS_MCP.RBLX.ENGAGEMENTS`, panel `All Games`, datasource `Consistent`: daily average concurrent players across our consistent panel of games, all platforms.

**Method:** both series are put on the same footing: a **7-day trailing mean (tr7d)** of the daily values, over the dates both sources cover (latest common date shown in section 1). Year-on-year (y/y) compares each tr7d with the tr7d 364 days earlier, and y/2y with the tr7d 728 days earlier (cumulative two-year growth, not annualised), so weekdays line up. Week-on-week compares each tr7d with the tr7d 7 days earlier, taken in non-overlapping steps back from the latest date so the newest days are always included.

### Takeaways
Data to 21 Oct 2026, the latest date both sources cover.
- **Q4 2026 hours, SRS apps method (section 12):** on Sensor Tower DAU the method gives **+5.3% y/y**, against **−0.5%** on Consistent CCU (the SRS apps figure, reproduced exactly) and **−10.2%** VA consensus. The gap comes from the Q2 2026 anchor. CCU's 2Y-implied y/y for Q2 2026 was +7.6%, above reported hours (+5.8%), while Sensor Tower's was −1.1%. So CCU says Q4 is running 6.4pp below the anchor on a two-year basis, and Sensor Tower says only 0.6pp below. On past quarters the CCU version has been far more accurate: mean absolute error 3.1pp against 9.2pp for Sensor Tower and 10.0pp for consensus (Q2 2023 to Q2 2026). Sensor Tower missed by about 20pp in both directions in 2025–26, as hours per user swung. Treat the CCU figure as the main estimate and Sensor Tower's as an upper bound.
- **Week to week, the two series mostly move together.** 7-day changes in the tr7d series have a correlation of 0.78 YTD, with a standard deviation of 3.2pp in the gap. Two steps fall beyond ±2 sd (about 6.4pp): the week to **14 Jan** (CCU −12.8% vs DAU −6.0%, the end of the holidays) and the week to **22 Apr** (CCU −6.2% vs DAU +1.0%, −7.1pp). The week to **1 Jul** (+6.5pp, start of US summer holidays) is right at the threshold. The week to **26 Aug**, straight after the clampdown, was −5.2pp (CCU −5.9% vs DAU −0.6%).
- **The clampdown shows up as a CCU drop with no matching DAU drop.** From 16 to 26 Aug, CCU tr7d fell 6.2% (12.34M to 11.57M) while DAU tr7d barely moved (166.1M to 165.0M). CCU per 1,000 DAU fell from 74.3 to 70.1.
- **It is not just back-to-school.** On the same calendar, CCU per 1,000 DAU barely moved in the 10 days after 16 Aug in 2024 (−1.2%) and 2025 (+0.6%), against −5.6% in 2026. It also fell further and has not recovered: by 7 Oct it was −15.6% in 2026 against −9.9% (2024) and −4.6% (2025), and by 21 Oct −13.1% against −3.1% (2025). So about **6–11pp of the 2026 fall is beyond the usual seasonal pattern**. That fits the removal of bot or AFK sessions that counted as concurrent players but not as app users.
- **The year-on-year gap widened after the clampdown.** CCU growth was already below DAU growth before it (gap about −6pp on 12–19 Aug). The gap went to −10.2pp by 26 Aug, −10 to −14pp in September and −16 to −20pp in October. On 21 Oct DAU is **+8.5% y/y** while CCU is **−7.6%**.
- **DAU growth: y/y is weak because of the 2025 base, y/2y is strong.** DAU y/y slowed from +35% (January average) to −6% (August), then recovered to +7% in October (+8.5% on 21 Oct). Over two years, growth held at about +52–57% from January to August, then jumped to +78% (September) and +84% (October), because Sep–Oct 2024 was a weak base. CCU y/2y is similar in Sep–Oct (+87%, +86%). So on a two-year view CCU and DAU have grown about the same, and the large y/y gap mostly reflects 2025 CCU running unusually high relative to users. CCU per 1,000 DAU was 78.6 on 16 Aug 2025, against 69.4 a year earlier and about 63–65 now. That fits the clampdown removing 2025-era bot inflation and bringing CCU per user back below 2024 levels. On the y/2y chart, CCU ran about 5–13pp ahead of DAU in August and September, then narrowed to about 0–4pp in October.
- **CCU per user was falling before August too.** It went from 89.1 (4 Jan) to about 65–74 from May to August, and has been about 62–65 since late September. Part of the early-year fall is the holiday peak unwinding. But the week to 22 Apr (CCU −6% with DAU up 1%) looks like an earlier clampdown-style drop, and is worth checking against any Roblox enforcement news from mid-April.
- **Backtest against Roblox's reported KPIs (section 11, Q1 2019 to Q2 2026):** Sensor Tower has tracked reported DAU and hours well over the last 5 quarters, beating consensus on average. Over the full history it is noisier than consensus, with large misses in 2020–21 and mid-2023. For Q3 2026 it points to DAU y/y of −2% to −4% and hours of about −8%, against consensus of −12.9% and −13.8%. The hours figure is likely an upper bound, because hours per user has been falling.
- **What this can't tell apart:** CCU ÷ DAU is mostly hours played per user. The timing points to the clampdown, but a fall in real hours, or a shift in play from mobile to PC/console (which DAU can't see), would look the same. Splitting the CCU drop by hour of day, or by the games known to host bots, would test it further.


**Caveats:** Sensor Tower DAU are modelled estimates, summed across iOS and Android, so a user on both counts twice. CCU ÷ DAU mostly moves with hours played per user, so it can't separate bots from real changes in time spent, or from shifts between mobile and PC/console. The CCU history is missing 13 days (25 Mar, 1 Apr, 18 Apr, 24 Apr, 9 Jul and 16–21 Oct 2024; 15 Apr and 27 Jun 2025). tr7d windows touching those days are dropped, which leaves short breaks in the CCU y/y line (mid-April, late June) and in the CCU y/2y line (late March to April, early July and mid-October). DAU has no gaps. DAU growth is computed from the DAU series alone, so it has no breaks.

```python
from datetime import date

import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import polars as pl
import snowflake.connector

DAU_FROM = "2018-01-01"  # Sensor Tower history pulled (the backtest in section 11 needs 2018 as a base)
YTD_START = pl.date(2026, 1, 1)
CLAMPDOWN = (pl.date(2026, 8, 17), pl.date(2026, 8, 19))  # bot clampdown, inclusive
FLAG_SD = 2  # flag 7-day steps whose CCU-minus-DAU growth gap is beyond this many standard deviations

# Validated reference palette (dataviz skill): categorical slots 1-2, muted ink for context.
BLUE, ORANGE, INK, MUTED, SHADE = "#2a78d6", "#eb6834", "#3d3d3a", "#898781", "#c3c2b7"
pl.Config.set_tbl_rows(60)
pl.Config.set_tbl_cols(-1)  # show every column
pl.Config.set_tbl_width_chars(200)
plt.rcParams.update({"figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
                     "axes.grid": True, "grid.color": "#e1e0d9", "grid.linewidth": 0.8})
```

## 1. Load the data

Both series come from Snowflake. Sensor Tower is summed over apps and devices for the worldwide row; the table below shows what goes into it.

```python
with snowflake.connector.connect(connection_name="default") as con:
    cur = con.cursor()
    # Worldwide row only: the per-country rows add up to it.
    cur.execute(f"""
        SELECT date::date AS date, product_name, device_code, SUM(dau) AS dau
        FROM sensortower.common.core
        WHERE unified_product_name ILIKE 'roblox'
            AND product_name IN ('Roblox', 'Roblox VN')
            AND country_code = 'WW'
            AND date >= '{DAU_FROM}'
        GROUP BY ALL
    """)
    st = pl.DataFrame(cur.fetchall(), schema=["date", "app", "device", "dau"], orient="row")

st = st.with_columns(pl.col("date").cast(pl.Date), pl.col("dau").cast(pl.Float64))
dau = st.group_by("date").agg(pl.col("dau").sum()).sort("date")

print(f"Sensor Tower DAU {dau['date'].min()} to {dau['date'].max()}")
st.group_by("app", "device").agg(
    first=pl.col("date").min(), last=pl.col("date").max(), days=pl.len(), avg_dau_m=(pl.col("dau").mean() / 1e6).round(1)
).sort("app", "device")
```

```python
with snowflake.connector.connect(connection_name="default") as con:
    cur = con.cursor()
    cur.execute(f"""
        SELECT date, players
        FROM srs_mcp.rblx.engagements
        WHERE panel = 'All Games' AND datasource = 'Consistent' AND date >= '{DAU_FROM}'
        ORDER BY date
    """)
    ccu = pl.DataFrame(cur.fetchall(), schema=["date", "ccu"], orient="row")

ccu = ccu.with_columns(pl.col("date").cast(pl.Date), pl.col("ccu").cast(pl.Float64))
print(f"CCU {ccu['date'].min()} to {ccu['date'].max()}")
```

## 2. Daily tr7d for both, plus growth rates

`rolling_mean_by` uses a true 7-day date window, so the days missing from the CCU history (see caveats) don't shift the window; windows with fewer than 7 days are dropped.

```python
def tr7d(df: pl.DataFrame, col: str) -> pl.DataFrame:
    """Adds a 7-day trailing mean of `col` and the number of days behind it (7 = complete window)."""
    return df.sort("date").with_columns(
        pl.col(col).rolling_mean_by("date", window_size="7d").alias(f"{col}_tr7d"),
        pl.col(col).is_not_null().cast(pl.Int32).rolling_sum_by("date", window_size="7d").alias(f"{col}_days"),
    )


def with_growth(df: pl.DataFrame, col: str) -> pl.DataFrame:
    """Keeps complete tr7d windows of one source and adds its growth against the same weekday
    1 year (364 days) and 2 years (728 days) earlier. Done per source, so a gap in one doesn't blank the other."""
    df = df.filter(pl.col(f"{col}_days") == 7).drop(f"{col}_days")
    tr = f"{col}_tr7d"
    for label, days in (("yoy", 364), ("y2y", 728)):
        past = df.select(pl.col("date").dt.offset_by(f"{days}d"), pl.col(tr).alias("_past"))
        df = df.join(past, on="date", how="left").with_columns(
            (pl.col(tr) / pl.col("_past") - 1).alias(f"{col}_{label}")
        ).drop("_past")
    return df


daily = (
    with_growth(tr7d(dau, "dau"), "dau")
    .join(with_growth(tr7d(ccu, "ccu"), "ccu"), on="date", how="inner")  # dates both sources cover
    .with_columns(
        ccu_per_1k_dau=pl.col("ccu_tr7d") / pl.col("dau_tr7d") * 1_000,
        yoy_gap=pl.col("ccu_yoy") - pl.col("dau_yoy"),
    )
)

LATEST = daily["date"].max()
ytd = daily.filter(pl.col("date") >= YTD_START)
print(f"Latest common date: {LATEST}")
ytd.tail(3)
```

**7-day steps.** Every 7th day counting back from the latest date, each tr7d is compared with the one 7 days before, so each step covers a separate week and the last step ends on the latest date. The gap is CCU growth minus DAU growth.

```python
steps = (
    daily.filter(((LATEST - pl.col("date")).dt.total_days() % 7) == 0)
    .with_columns(
        dau_7d=pl.col("dau_tr7d").pct_change(),
        ccu_7d=pl.col("ccu_tr7d").pct_change(),
    )
    .with_columns(gap_7d=pl.col("ccu_7d") - pl.col("dau_7d"))
    .filter(pl.col("date") >= YTD_START)
)
gap_sd = steps["gap_7d"].std()
steps = steps.with_columns(flag=pl.col("gap_7d").abs() > FLAG_SD * gap_sd)
print(f"{steps.height} steps, last ends {steps['date'].max()}; gap sd {gap_sd * 100:.1f}pp")
```

## 3. Levels: tr7d CCU and tr7d DAU

```python
def shade_clampdown(ax, label=True):
    """Shades the 17-19 Aug bot clampdown (to the end of 19 Aug)."""
    start, end = (pl.select(d).item() for d in CLAMPDOWN)
    ax.axvspan(start, end + pl.select(pl.duration(days=1)).item(), color=SHADE, alpha=0.45, lw=0, zorder=0)
    if label:
        ax.text(start, 1.0, " Bot Clampdown", transform=ax.get_xaxis_transform(), ha="left", va="top",
                fontsize=8.5, color=INK, fontweight="bold")


def month_axis(ax):
    ax.xaxis.set_major_locator(mdates.MonthLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))


def levels_chart(df: pl.DataFrame, year: int, clampdown: bool):
    """Two panels for one calendar year: tr7d levels on twin axes, and both indexed to the year's first tr7d = 100."""
    base = df.row(0, named=True)
    fig, (ax1, ax3) = plt.subplots(1, 2, figsize=(14, 4.8))

    ax1.plot(df["date"], df["ccu_tr7d"] / 1e6, color=BLUE, lw=2, label="CCU, tr7d (left, M)")
    ax1.set_ylabel("Consistent CCU (M)", color=BLUE)
    ax2 = ax1.twinx()
    ax2.plot(df["date"], df["dau_tr7d"] / 1e6, color=ORANGE, lw=2, label="DAU, tr7d (right, M)")
    ax2.set_ylabel("Sensor Tower DAU (M)", color=ORANGE)
    ax2.grid(False)
    ax2.spines["right"].set_visible(True)
    lines = ax1.get_legend_handles_labels()[0] + ax2.get_legend_handles_labels()[0]
    ax1.legend(lines, [l.get_label() for l in lines], loc="lower right", frameon=False, fontsize=9)
    ax1.set_title(f"Levels (tr7d), {year}")

    ax3.plot(df["date"], df["ccu_tr7d"] / base["ccu_tr7d"] * 100, color=BLUE, lw=2, label="CCU")
    ax3.plot(df["date"], df["dau_tr7d"] / base["dau_tr7d"] * 100, color=ORANGE, lw=2, label="DAU")
    ax3.axhline(100, color=MUTED, lw=1, ls="--")
    ax3.set_title(f"Indexed, tr7d on {base['date']:%d %b %Y} = 100")
    ax3.legend(loc="lower right", frameon=False, fontsize=9)

    for ax in (ax1, ax3):
        if clampdown:
            shade_clampdown(ax)
        month_axis(ax)
    fig.tight_layout()


levels_chart(ytd, 2026, clampdown=True)
```

The same two charts for **2025** (1 Jan to 31 Dec), for comparison. There is no clampdown band, because the clampdown was in 2026. Our CCU history is missing 15 Apr and 27 Jun 2025, so the CCU line bridges those short gaps.

```python
levels_chart(daily.filter(pl.col("date").dt.year() == 2025), 2025, clampdown=False)
```

## 4. Week-on-week: moves in one but not the other

Top: 7-day change in each tr7d. Bottom: the gap (CCU growth minus DAU growth); dark bars are beyond ±2 standard deviations. A positive gap means CCU rose more than DAU (more time per user, or more non-human sessions). A negative gap means the reverse.

```python
fig, (top, bot) = plt.subplots(2, 1, figsize=(14, 7), sharex=True, gridspec_kw={"height_ratios": [3, 2]})
top.plot(steps["date"], steps["ccu_7d"] * 100, color=BLUE, lw=2, marker="o", ms=3, label="CCU tr7d, 7-day change")
top.plot(steps["date"], steps["dau_7d"] * 100, color=ORANGE, lw=2, marker="o", ms=3, label="DAU tr7d, 7-day change")
top.axhline(0, color=MUTED, lw=1)
top.set_ylabel("%")
top.legend(frameon=False, fontsize=9)
corr = steps.select(pl.corr("ccu_7d", "dau_7d")).item()
top.set_title(f"7-day change (correlation {corr:.2f})")

colors = [INK if f else MUTED for f in steps["flag"]]
bot.bar(steps["date"], steps["gap_7d"] * 100, width=5, color=colors)
for sign in (1, -1):
    bot.axhline(sign * FLAG_SD * gap_sd * 100, color=MUTED, lw=1, ls="--")
bot.axhline(0, color=MUTED, lw=1)
bot.set_ylabel("CCU minus DAU (pp)")
bot.set_title(f"Gap; dark bars beyond ±{FLAG_SD} sd ({FLAG_SD * gap_sd * 100:.1f}pp)")

shade_clampdown(top)
shade_clampdown(bot, label=False)
month_axis(bot)
fig.tight_layout()

steps.filter("flag").select(
    pl.col("date").alias("week_ending"),
    (pl.col("dau_7d", "ccu_7d", "gap_7d") * 100).round(1).name.suffix("_pct"),
)
```

## 5. Growth: Sensor Tower DAU vs Consistent CCU, y/y and y/2y

Both series' tr7d against the same weekday one year (364 days) and two years (728 days) earlier. Growth removes most of the seasonality behind the weekly gaps. The y/2y view is total growth over two years (not annualised), so it shows whether 2026 is up on 2024 as well as on 2025, which had an unusually strong summer. The shaded area is the gap (CCU minus DAU).

```python
def latest(col: str) -> str:
    """Latest non-null value of a growth column, with its date (the CCU y/2y base has a gap in Oct 2024)."""
    row = ytd.filter(pl.col(col).is_not_null()).row(-1, named=True)
    return f"{row[col]:+.1%} ({row['date']:%d %b})"


fig, axes = plt.subplots(2, 1, figsize=(14, 8.5), sharex=True)

for ax, (key, title) in zip(axes, (("yoy", "y/y (vs 364 days earlier)"), ("y2y", "y/2y (vs 728 days earlier, cumulative)"))):
    ccu_g, dau_g = ytd[f"ccu_{key}"] * 100, ytd[f"dau_{key}"] * 100
    ax.plot(ytd["date"], ccu_g, color=BLUE, lw=2, label="Consistent CCU")
    ax.plot(ytd["date"], dau_g, color=ORANGE, lw=2, label="Sensor Tower DAU")
    ax.fill_between(ytd["date"], ccu_g - dau_g, 0, color=MUTED, alpha=0.3, lw=0, label="Gap (CCU minus DAU, pp)")
    ax.axhline(0, color=MUTED, lw=1)
    ax.set_ylabel("%")
    ax.set_title(f"{title}; latest CCU {latest(f'ccu_{key}')}, DAU {latest(f'dau_{key}')}")
    ax.legend(loc="lower left", frameon=False, fontsize=9)
    shade_clampdown(ax)
month_axis(axes[-1])
fig.tight_layout()

# Monthly means of the daily tr7d growth rates; October runs to the latest date.
(
    ytd.group_by(month=pl.col("date").dt.truncate("1mo"))
    .agg((pl.col("dau_yoy", "ccu_yoy", "dau_y2y", "ccu_y2y").mean() * 100).round(1))
    .sort("month")
    .with_columns(pl.col("month").dt.strftime("%b %Y"))
)
```

## 6. Since 1 October: y/y and y/2y, day by day

The same two growth views from 1 October to the latest date, with every day labelled. Each point is the tr7d growth for the 7 days ending that day. CCU y/2y is blank from 14 Oct: its 2024 base windows include the missing 16–21 Oct 2024 CCU days, so it returns only on 28 Oct.

```python
recent = ytd.filter(pl.col("date") >= pl.date(2026, 10, 1))

fig, axes = plt.subplots(2, 1, figsize=(14, 8.5), sharex=True)
for ax, (key, title) in zip(axes, (("yoy", "y/y (vs 364 days earlier)"), ("y2y", "y/2y (vs 728 days earlier, cumulative)"))):
    for col, other, color, name in (("ccu", "dau", BLUE, "Consistent CCU"), ("dau", "ccu", ORANGE, "Sensor Tower DAU")):
        g, g_other = recent[f"{col}_{key}"] * 100, recent[f"{other}_{key}"] * 100
        ax.plot(recent["date"], g, color=color, lw=2, marker="o", ms=4, label=name)
        for d, v, w in zip(recent["date"], g, g_other):  # label each point; the higher series above, the lower below
            if v is not None:
                above = w is None or v >= w
                ax.annotate(f"{v:.1f}", (d, v), textcoords="offset points", xytext=(0, 7 if above else -13),
                            ha="center", fontsize=7.5, color=color)
    if key == "yoy":
        ax.axhline(0, color=MUTED, lw=1)  # y/2y stays well above zero, so no zero line there
    gap = recent.filter(pl.col(f"ccu_{key}").is_null())
    if gap.height:
        ax.text(gap["date"].min(), 0.04, f"  CCU {title.split()[0]} unavailable from {gap['date'].min():%d %b}: its base window has missing CCU days",
                transform=ax.get_xaxis_transform(), fontsize=8.5, color=BLUE)
    ax.set_ylabel("%")
    ax.set_title(title)
    ax.legend(loc="center left", frameon=False, fontsize=9)
    ax.margins(y=0.15)

first, last_day = (mdates.date2num(d) for d in (recent["date"].min(), recent["date"].max()))
axes[-1].set_xlim(first - 0.75, last_day + 0.75)  # room for the end labels
axes[-1].xaxis.set_major_locator(mdates.DayLocator())
axes[-1].xaxis.set_major_formatter(mdates.DateFormatter("%a %d %b"))
plt.setp(axes[-1].get_xticklabels(), fontsize=8, rotation=90)  # one label per day
for ax in axes:
    ax.grid(axis="x", color="#e1e0d9", lw=0.6)
fig.tight_layout()

recent.select(
    "date",
    pl.col("date").dt.strftime("%a").alias("day"),
    (pl.col("dau_yoy", "ccu_yoy", "dau_y2y", "ccu_y2y") * 100).round(1),
)
```

## 9. Is it just back-to-school? The same window in 2024 and 2025

CCU per 1,000 DAU (tr7d) from 1 Aug to 31 Oct each year, indexed to 16 Aug = 100. Back-to-school cuts hours per user every year, so the question is whether 2026 fell earlier or further than the seasonal pattern.

## 11. Backtest: Sensor Tower DAU vs Roblox's reported DAU and hours (y/y)

Roblox reports **DAU** (quarterly average of daily active users) and **hours engaged** (quarterly total). Both come from Visible Alpha (`SRS_MCP.VISIBLE_ALPHA.KPIS`, company 28412, KPIs 14113119 and 14113120), reported from Q1 2018.

It uses the same Sensor Tower series as the rest of the notebook (Snowflake `WW` row, from 2018), so the backtest runs from Q1 2019. Only quarters with every day present are used.

**Estimates, each on a y/y basis**
- **DAU, direct:** Sensor Tower average daily DAU in the quarter against the same quarter a year earlier.
- **DAU, sequential:** last quarter's reported DAU y/y plus the change (pp) in Sensor Tower's average-DAU y/y since last quarter.
- **Hours, sequential (the model):**
  1. Sum Sensor Tower DAU over the quarter.
  2. y/y = summed DAU in quarter t ÷ summed DAU in quarter t−4, minus 1.
  3. Take the change in that y/y against the previous quarter (pp).
  4. Add the change to the previous quarter's reported hours y/y.

  Summing (rather than averaging) keeps leap days in the base, as they are in reported hours.

Q3 2026 has full Sensor Tower data but is not reported yet, so it is shown as an **estimate** next to the current Visible Alpha consensus. For quarters already reported, Visible Alpha's `consensus_value` is the last consensus it holds, not a point-in-time pre-earnings snapshot. Consensus errors are therefore a rough guide only.

```python
VA_COMPANY = 28412
VA_KPIS = {14113119: "dau", 14113120: "hours"}  # DAU (quarterly average), Hours engaged (quarterly total)
BACKTEST_FROM = pl.date(2019, 1, 1)  # first quarter with a reported y/y (Visible Alpha starts Q1 2018)

with snowflake.connector.connect(connection_name="default") as con:
    cur = con.cursor()
    cur.execute(f"""
        SELECT DATE_TRUNC('quarter', period_end_dt) AS quarter,  -- VA start dates can sit a day or two before the quarter
            kpi_id, reported_value, consensus_value
        FROM srs_mcp.visible_alpha.kpis
        WHERE va_company_id = {VA_COMPANY}
            AND kpi_id IN ({", ".join(map(str, VA_KPIS))})
            AND period_type = 'QTR'
            AND period_end_dt BETWEEN '2018-03-31' AND '2026-12-31'
    """)
    va_raw = pl.DataFrame(cur.fetchall(), schema=["quarter", "kpi_id", "reported", "consensus"], orient="row")

va = (
    va_raw.with_columns(
        pl.col("quarter").cast(pl.Date),
        pl.col("kpi_id").replace_strict(VA_KPIS).alias("kpi"),
        pl.col("reported", "consensus").cast(pl.Float64),
    )
    .pivot(on="kpi", index="quarter", values=["reported", "consensus"])
    .sort("quarter")
    .with_columns(
        # Reported y/y; consensus y/y is against the reported value a year earlier.
        dau_rep_yy=pl.col("reported_dau") / pl.col("reported_dau").shift(4) - 1,
        hrs_rep_yy=pl.col("reported_hours") / pl.col("reported_hours").shift(4) - 1,
        dau_cons_yy=pl.col("consensus_dau") / pl.col("reported_dau").shift(4) - 1,
        hrs_cons_yy=pl.col("consensus_hours") / pl.col("reported_hours").shift(4) - 1,
    )
)
print(f"Visible Alpha quarters: {va['quarter'].min()} to {va['quarter'].max()}")
```

```python
def quarterly_estimates(daily_dau: pl.DataFrame) -> pl.DataFrame:
    """Sensor Tower y/y and the three estimates per calendar quarter, from a daily (date, dau) frame.
    Only quarters with every day present count, and the y/y needs the same quarter a year earlier."""
    q = (
        daily_dau.group_by(quarter=pl.col("date").dt.truncate("1q"))
        .agg(st_sum=pl.col("dau").sum(), st_avg=pl.col("dau").mean(), last=pl.col("date").max(), days=pl.len())
        .filter(pl.col("last") == pl.col("quarter").dt.offset_by("1q").dt.offset_by("-1d"))
        .drop("last")
    )
    # Join onto the full VA quarter list so shift(4) and shift(1) step by exactly one year / one quarter.
    return (
        va.join(q, on="quarter", how="left")
        .sort("quarter")
        .with_columns(
            st_avg_yy=pl.col("st_avg") / pl.col("st_avg").shift(4) - 1,
            st_sum_yy=pl.col("st_sum") / pl.col("st_sum").shift(4) - 1,
        )
        .with_columns(
            dau_est_direct=pl.col("st_avg_yy"),
            dau_est_seq=pl.col("dau_rep_yy").shift(1) + pl.col("st_avg_yy").diff(),
            hrs_est_seq=pl.col("hrs_rep_yy").shift(1) + pl.col("st_sum_yy").diff(),
        )
        .filter(pl.col("st_avg_yy").is_not_null() & (pl.col("quarter") >= BACKTEST_FROM))
        .with_columns(
            label=pl.format("Q{} {}", pl.col("quarter").dt.quarter(), pl.col("quarter").dt.year()),
            is_estimate=pl.col("reported_dau").is_null(),
        )
    )


bt = quarterly_estimates(dau)  # from Q1 2019

pct = ["dau_rep_yy", "dau_est_direct", "dau_est_seq", "dau_cons_yy", "hrs_rep_yy", "hrs_est_seq", "hrs_cons_yy"]
bt.select("label", "days", (pl.col(pct) * 100).round(1))
```

### Errors

Estimate minus reported (pp). *Direction* asks whether the hours model moved the same way as reported hours y/y against the previous quarter (acceleration vs deceleration). The first summary covers the Sensor Tower estimates over every quarter they exist. The second compares them with VA consensus like for like (consensus starts in Q1 2021), for the whole period and for recent windows.

```python
from IPython.display import display

ERRORS = {"dau_direct": ("dau_est_direct", "dau_rep_yy"), "dau_seq": ("dau_est_seq", "dau_rep_yy"),
          "dau_cons": ("dau_cons_yy", "dau_rep_yy"), "hrs_model": ("hrs_est_seq", "hrs_rep_yy"),
          "hrs_cons": ("hrs_cons_yy", "hrs_rep_yy")}


def error_table(df: pl.DataFrame) -> pl.DataFrame:
    """Per reported quarter: each estimate's error (pp) and whether the hours model called the direction right."""
    return df.filter(~pl.col("is_estimate")).select(
        "quarter", "label",
        *[((pl.col(est) - pl.col(rep)) * 100).alias(f"{name}_err") for name, (est, rep) in ERRORS.items()],
        hrs_direction_ok=((pl.col("hrs_est_seq") - pl.col("hrs_rep_yy").shift(1)).sign()
                          == (pl.col("hrs_rep_yy") - pl.col("hrs_rep_yy").shift(1)).sign()),
    )


def summarise(errs: pl.DataFrame) -> pl.DataFrame:
    """MAE, bias and hours direction hit rate over the quarters where every estimate (and consensus) exists."""
    common = errs.drop_nulls([f"{n}_err" for n in ERRORS])  # like-for-like quarters only
    period = f"{common['label'][0]} - {common['label'][-1]}"
    cols = [f"{n}_err" for n in ERRORS]
    base = [pl.lit(period).alias("period"), pl.len().alias("qtrs")]
    return pl.concat([
        common.select(*base, pl.lit("MAE").alias("stat"), *[pl.col(c).abs().mean().round(1) for c in cols],
                      (pl.col("hrs_direction_ok").mean() * 100).round(0).alias("hrs_direction_pct")),
        common.select(*base, pl.lit("Bias").alias("stat"), *[pl.col(c).mean().round(1) for c in cols],
                      pl.lit(None, dtype=pl.Float64).alias("hrs_direction_pct")),
    ])


err = error_table(bt)

# Sensor Tower estimates alone, every quarter they exist (VA consensus only starts in Q1 2021).
model_only = err.drop_nulls(["dau_seq_err", "hrs_model_err"])
model_summary = model_only.select(
    pl.lit(f"{model_only['label'][0]} - {model_only['label'][-1]}").alias("period"), pl.len().alias("qtrs"),
    *[pl.col(f"{n}_err").abs().mean().round(1).alias(f"{n}_mae") for n in ("dau_direct", "dau_seq", "hrs_model")],
    (pl.col("hrs_direction_ok").mean() * 100).round(0).alias("hrs_direction_pct"),
)

# Against consensus, like for like.
summary = pl.concat([
    summarise(err),
    summarise(err.filter(pl.col("quarter") >= pl.date(2023, 1, 1))),
    summarise(err.filter(pl.col("quarter") >= pl.date(2025, 4, 1))),
])
display(model_summary)
display(err.drop("quarter").with_columns(pl.col("^.*_err$").round(1)))
summary
```

## 12. Q4 2026 hours nowcast: the SRS apps method on CCU vs Sensor Tower DAU

This replicates the hours nowcast on apps.srsinvest.com/internet/roblox (`SRS_ANALYSTS.FINTAN_CREEDON.RBLX_V_OPEN_QUARTER_V2`) and runs it on two signals: our Consistent CCU (as the app does) and Sensor Tower DAU.

**Method (2-year-implied):**
- **2Y-implied y/y** for a quarter = reported hours 8 quarters earlier × (1 + the signal's 2-year growth) ÷ reported hours 4 quarters earlier − 1. In effect, the signal's 2-year stack is applied to the hours base two years back, then compared with last year's reported hours.
- **Estimate** = the anchor quarter's reported hours y/y + (open quarter's 2Y-implied y/y − anchor quarter's 2Y-implied y/y). The anchor is the last reported quarter, Q2 2026, because Q3 2026 is not reported yet.
- **The open quarter's 2-year growth (the seed):**
  - Under 28 days into the quarter, it is the latest 7 days against the same calendar dates two years earlier.
  - If either 7-day window has a missing day, or from day 28 onwards, it is quarter to date against the same dates two years earlier.
  - Each signal uses its own latest date (CCU 22 Oct, Sensor Tower 21 Oct). CCU falls back to quarter to date, because its 2024 base window includes the missing 16–21 Oct 2024 days.
- **1-year version (for reference):** anchor hours y/y + (quarter-to-date y/y − anchor quarter y/y).

The app's own figure is read from the view as a check.

```python
from datetime import timedelta

OPEN_Q, ANCHOR_Q = date(2026, 10, 1), date(2026, 4, 1)
SEED_7D_DAYS = 28  # under this many days into the quarter, seed on the latest 7 days
hours = dict(zip(va["quarter"], va["reported_hours"]))  # reported hours engaged by quarter start


def add_q(q: date, n: int) -> date:
    m = q.month - 1 + 3 * n
    return date(q.year + m // 12, m % 12 + 1, 1)


def q_end(q: date) -> date:
    return add_q(q, 1) - timedelta(days=1)


def years_back(d: date, n: int) -> date:
    return d.replace(year=d.year - n) if not (d.month == 2 and d.day == 29) else date(d.year - n, 2, 28)


def window(sig: pl.DataFrame, start: date, end: date) -> tuple[float, int]:
    """Mean of the signal over [start, end] and the number of days present."""
    w = sig.filter(pl.col("date").is_between(start, end))
    return w["value"].mean(), w.height


def implied_2y(sig: pl.DataFrame, q: date, growth: float | None = None) -> float:
    """2Y-implied hours y/y for quarter q: hours(q-8) x (1 + 2-year signal growth) / hours(q-4) - 1.
    The growth defaults to the full quarter against the same quarter two years earlier."""
    if growth is None:
        growth = window(sig, q, q_end(q))[0] / window(sig, add_q(q, -8), q_end(add_q(q, -8)))[0] - 1
    return hours[add_q(q, -8)] * (1 + growth) / hours[add_q(q, -4)] - 1


def nowcast(sig: pl.DataFrame, name: str) -> dict:
    cutoff = sig["date"].max()
    days_in = (cutoff - OPEN_Q).days + 1
    m7, n7 = window(sig, cutoff - timedelta(days=6), cutoff)
    b7, n7_base = window(sig, years_back(cutoff, 2) - timedelta(days=6), years_back(cutoff, 2))
    mq, _ = window(sig, OPEN_Q, cutoff)
    b2q, _ = window(sig, years_back(OPEN_Q, 2), years_back(cutoff, 2))
    b1q, _ = window(sig, years_back(OPEN_Q, 1), years_back(cutoff, 1))
    use_7d = days_in < SEED_7D_DAYS and n7 == 7 and n7_base == 7
    seed = m7 / b7 - 1 if use_7d else mq / b2q - 1

    anchor_rep = hours[ANCHOR_Q] / hours[add_q(ANCHOR_Q, -4)] - 1
    anchor_2y = implied_2y(sig, ANCHOR_Q)
    open_2y = implied_2y(sig, OPEN_Q, growth=seed)
    anchor_1y = window(sig, ANCHOR_Q, q_end(ANCHOR_Q))[0] / window(sig, add_q(ANCHOR_Q, -4), q_end(add_q(ANCHOR_Q, -4)))[0] - 1
    return {
        "signal": name, "data_to": cutoff, "days_in": days_in, "seed": "7d" if use_7d else "qtd",
        "seed_2y_growth": seed, "anchor_hours_yy": anchor_rep, "anchor_2y_implied": anchor_2y,
        "open_2y_implied": open_2y, "hours_est_2y": anchor_rep + open_2y - anchor_2y,
        "hours_est_1y": anchor_rep + (mq / b1q - 1) - anchor_1y,
    }


signals = {
    "Consistent CCU": ccu.rename({"ccu": "value"}),
    "Sensor Tower DAU": dau.rename({"dau": "value"}),
}
nc = pl.DataFrame([nowcast(sig, name) for name, sig in signals.items()])

with snowflake.connector.connect(connection_name="default") as con:
    cur = con.cursor()
    cur.execute("""
        SELECT hours_est_yy, hours_est_yy_1y, hours_cons_yy, as_of_date
        FROM srs_analysts.fintan_creedon.rblx_v_open_quarter_v2
    """)
    app_est, app_est_1y, hours_cons, app_asof = cur.fetchone()

pct_cols = ["seed_2y_growth", "anchor_hours_yy", "anchor_2y_implied", "open_2y_implied", "hours_est_2y", "hours_est_1y"]
print(f"SRS apps view (as of {app_asof}): 2Y {float(app_est):+.2%}, 1Y {float(app_est_1y):+.2%}; "
      f"VA consensus (point in time) {float(hours_cons):+.2%}")
nc.with_columns((pl.col(pct_cols) * 100).round(2))
```

### How has this method done? Full-quarter backtest, CCU vs Sensor Tower

The same 2Y-implied method applied to each completed quarter with the whole quarter's data: estimate = previous quarter's reported hours y/y + (this quarter's 2Y-implied y/y − previous quarter's). This is the end-of-quarter version of the nowcast, so it shows each signal's best case. Consistent CCU history is too short for a two-year base before then, so its estimates begin in Q2 2023. The last point, Q3 2026, is not reported yet: there the method gives about −12% on CCU and −8% on Sensor Tower, against −13.8% consensus.

```python
def method_backtest(sig: pl.DataFrame) -> dict[date, float]:
    """End-of-quarter 2Y-implied hours estimate for each quarter where the signal covers both quarters and their bases."""
    first = sig["date"].min()
    out = {}
    for q in sorted(hours):
        prev = add_q(q, -1)
        needed = (prev, add_q(q, -4), add_q(q, -8), add_q(prev, -4), add_q(prev, -8))
        if q_end(q) > sig["date"].max() or add_q(prev, -8) < first or any(hours.get(n) is None for n in needed):
            continue
        prev_rep = hours[prev] / hours[add_q(prev, -4)] - 1
        out[q] = prev_rep + implied_2y(sig, q) - implied_2y(sig, prev)
    return out


method = (
    va.select("quarter", "hrs_rep_yy", "hrs_cons_yy")
    .with_columns(
        pl.col("quarter").replace_strict(method_backtest(sig), default=None, return_dtype=pl.Float64).alias(col)
        for col, sig in (("ccu_est", signals["Consistent CCU"]), ("st_est", signals["Sensor Tower DAU"]))
    )
    .filter(pl.col("st_est").is_not_null() | pl.col("ccu_est").is_not_null())
    .with_columns(label=pl.format("Q{} {}", pl.col("quarter").dt.quarter(), pl.col("quarter").dt.year()))
    .with_columns(
        ccu_err=pl.col("ccu_est") - pl.col("hrs_rep_yy"),
        st_err=pl.col("st_est") - pl.col("hrs_rep_yy"),
        cons_err=pl.col("hrs_cons_yy") - pl.col("hrs_rep_yy"),
    )
)

shown = method.filter(pl.col("quarter") >= pl.date(2021, 1, 1))
fig, ax = plt.subplots(figsize=(14, 5))
x = list(range(shown.height))
ax.plot(x, shown["hrs_rep_yy"] * 100, color=INK, lw=2.6, marker="o", ms=5, label="Reported hours")
ax.plot(x, shown["ccu_est"] * 100, color=BLUE, lw=1.8, marker="o", ms=4, label="Method on Consistent CCU")
ax.plot(x, shown["st_est"] * 100, color=ORANGE, lw=1.8, marker="o", ms=4, label="Method on Sensor Tower DAU")
ax.plot(x, shown["hrs_cons_yy"] * 100, color=MUTED, lw=1.4, ls=":", label="VA consensus")
ax.axhline(0, color=MUTED, lw=1)
ax.set_xticks(x, shown["label"], rotation=45, ha="right")
ax.set_ylabel("y/y %")
ax.set_title("Hours engaged y/y: reported vs the SRS apps method, end of quarter")
ax.legend(frameon=False, fontsize=9)
fig.tight_layout()

common = method.drop_nulls(["ccu_err", "st_err"])
display(common.select("label", (pl.col("hrs_rep_yy", "ccu_est", "st_est", "hrs_cons_yy", "ccu_err", "st_err", "cons_err") * 100).round(1)))
common.select(
    pl.lit(f"{common['label'][0]} - {common['label'][-1]}").alias("period"), pl.len().alias("qtrs"),
    *[pl.col(c).abs().mean().mul(100).round(1).alias(c.replace("_err", "_mae")) for c in ("ccu_err", "st_err", "cons_err")],
)
```

## 13. Reported DAU vs Sensor Tower average DAU, by quarter

Roblox's reported DAU (Visible Alpha) against Sensor Tower's average daily DAU for the same calendar quarter (worldwide row, Roblox + Roblox VN, iOS + Android). Only quarters with every day of Sensor Tower data are shown. Q3 2026 is not reported yet, so only Sensor Tower appears there.
- **Nominal:** quarterly average DAU in millions. The two tracked almost exactly through 2020. Since 2021, Sensor Tower's modelled total (mobile only) has run above Roblox's reported figure: by about 2–22M a day, and by more than 20M in Q3 2023, Q3 2025 and Q2 2026. So the levels are not directly comparable.
- **y/y:** each quarter against the same quarter a year earlier.

```python
dau_q = (
    dau.group_by(quarter=pl.col("date").dt.truncate("1q"))
    .agg(st_avg=pl.col("dau").mean(), last=pl.col("date").max())
    .filter(pl.col("last") == pl.col("quarter").dt.offset_by("1q").dt.offset_by("-1d"))  # full quarters only
    .drop("last")
)
qtr = (
    va.select("quarter", "reported_dau")
    .join(dau_q, on="quarter", how="full", coalesce=True)
    .sort("quarter")
    .with_columns(
        rep_yy=pl.col("reported_dau") / pl.col("reported_dau").shift(4) - 1,
        st_yy=pl.col("st_avg") / pl.col("st_avg").shift(4) - 1,
    )
    .filter(pl.col("st_avg").is_not_null() & (pl.col("quarter") >= pl.date(2018, 1, 1)))
    .with_columns(label=pl.format("Q{} {}", pl.col("quarter").dt.quarter(), pl.col("quarter").dt.year()))
)


def quarter_lines(ax, df, cols, title, ylabel, fmt):
    """Reported (ink) vs Sensor Tower (orange) by quarter, with the latest value of each labelled."""
    x = list(range(df.height))
    for col, color, name in cols:
        y = df[col]
        ax.plot(x, y, color=color, lw=2.2, marker="o", ms=4, label=name)
        last = max(i for i, v in enumerate(y) if v is not None)
        ax.annotate(fmt(y[last]), (last, y[last]), textcoords="offset points", xytext=(8, 0), va="center",
                    fontsize=8.5, color=color)
    ax.set_xticks(x[::2], df["label"][::2], rotation=45, ha="right")
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    ax.legend(frameon=False, fontsize=9, loc="upper left")


fig, ax = plt.subplots(figsize=(14, 5))
quarter_lines(ax, qtr.with_columns(pl.col("reported_dau", "st_avg") / 1e6),
              [("reported_dau", INK, "Roblox reported DAU"), ("st_avg", ORANGE, "Sensor Tower average DAU")],
              "Quarterly average DAU (nominal, millions)", "DAU (M)", lambda v: f"{v:.1f}M")
fig.tight_layout()
```
